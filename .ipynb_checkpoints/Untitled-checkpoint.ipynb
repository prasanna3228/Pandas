```python
import pandas as pd

Sheet1=pd.DataFrame({
    "id":['A1','A2','A5'],
    "name":['virat','patidar','venkat']
})
Sheet2=pd.DataFrame({
    "id":['A1','A2','A5',"a7"],
    "Salary":[900000,700000,500000,77000]
})

```

```python
Mergin_Two_Sheets=pd.merge(Sheet1,Sheet2,on="id",how="join")
print(Mergin_Two_Sheets)
```
